```python
import boto3
```

```python
from botocore.exceptions import ClientError

s3 = boto3.client('s3')

bucket_name = 'leoquiroga0728-bucket-2026'
archivo_local = 'archivo_prueba.txt'
archivo_s3 = 'archivo_prueba_boto3.txt'

# Verificar si el bucket existe
try:
    s3.head_bucket(Bucket=bucket_name)
    print("El bucket ya existe")
except ClientError:
    print("El bucket no existe, creando bucket...")
    s3.create_bucket(Bucket=bucket_name)
    print("Bucket creado correctamente")

# Subir archivo
try:
    s3.upload_file(archivo_local, bucket_name, archivo_s3)
    print("Archivo subido correctamente")
except ClientError as e:
    print("Error al subir archivo:", e)
```
